```python
# ============================================================
# ALLANAI.CrimeVision — Complete UK Crime Dataset Builder
# Author: Mahira
# Source: https://data.police.uk/api/
# ============================================================

import os
import time
import json
import requests
import pandas as pd
from pathlib import Path
from tqdm import tqdm

# --- Paths ---
ROOT = Path("../data")
RAW_DIR = ROOT / "raw"
INTERIM_DIR = ROOT / "interim"
PROC_DIR = ROOT / "processed"
for d in [RAW_DIR, INTERIM_DIR, PROC_DIR]:
    d.mkdir(parents=True, exist_ok=True)

# --- API Base ---
BASE_URL = "https://data.police.uk/api"

# ============================================================
# 1️ Helper functions
# ============================================================
def get_json(url, params=None):
    r = requests.get(url, params=params, timeout=60)
    r.raise_for_status()
    return r.json()

def list_forces():
    return get_json(f"{BASE_URL}/forces")

def list_force_neighbourhoods(force_id):
    return get_json(f"{BASE_URL}/{force_id}/neighbourhoods")

def get_neighbourhood_boundary(force_id, neigh_id):
    return get_json(f"{BASE_URL}/{force_id}/{neigh_id}/boundary")

def fetch_crimes_by_centroid(force_id, neigh_id, boundary_points, month):
    """Fetch crimes for one neighbourhood using centroid (safe for long URLs)."""
    if not boundary_points:
        return pd.DataFrame()

    lat = sum(float(p['latitude']) for p in boundary_points) / len(boundary_points)
    lng = sum(float(p['longitude']) for p in boundary_points) / len(boundary_points)

    params = {"lat": lat, "lng": lng, "date": month}
    url = f"{BASE_URL}/crimes-street/all-crime"
    try:
        data = get_json(url, params=params)
    except Exception:
        return pd.DataFrame()
    if not data:
        return pd.DataFrame()

    df = pd.json_normalize(data)
    cols = {
        "id": "crime_id",
        "category": "crime_type",
        "month": "month",
        "location.latitude": "latitude",
        "location.longitude": "longitude",
        "location.street.name": "street_name",
        "outcome_status.category": "outcome"
    }
    for c in cols.keys():
        if c not in df.columns:
            df[c] = pd.NA
    df = df[list(cols.keys())].rename(columns=cols)
    df["force_id"] = force_id
    df["neighbourhood_id"] = neigh_id
    return df

# ============================================================
# 2️ Fetch all forces and prepare month list
# ============================================================
forces = list_forces()
forces_df = pd.DataFrame(forces)
forces_df.head()
print(f"Found {len(forces_df)} police forces")

# Months available: from 2011-01 to 2024-12 (you can adjust)
import datetime as dt
months = pd.date_range(dt.date(2011,1,1), dt.date(2024,12,1), freq="MS").strftime("%Y-%m").tolist()

# ============================================================
# 3️ Fetch loop
# ============================================================
def fetch_force_month(force_id, month, sleep=1):
    neighs = list_force_neighbourhoods(force_id)
    if not neighs:
        return None
    all_dfs = []
    for n in neighs:
        try:
            boundary = get_neighbourhood_boundary(force_id, n["id"])
            df = fetch_crimes_by_centroid(force_id, n["id"], boundary, month)
            if not df.empty:
                all_dfs.append(df)
            time.sleep(sleep)
        except Exception as e:
            print(f"[WARN] {force_id}-{n['id']} {month}: {e}")
            time.sleep(sleep)
            continue

    if not all_dfs:
        return None
    df = pd.concat(all_dfs, ignore_index=True)
    out_path = INTERIM_DIR / f"{month}_{force_id}.csv"
    df.to_csv(out_path, index=False)
    return out_path

# ============================================================
# 4️ Master loop for all years
# ============================================================
# Start small for testing
# forces_subset = ["metropolitan", "west-midlands"]
forces_subset = [f["id"] for f in forces]

for month in months:
    print(f"\n📅 {month}")
    for fid in tqdm(forces_subset, desc=f"Fetching {month}"):
        try:
            fetch_force_month(fid, month)
        except Exception as e:
            print(f"[ERR] {fid} {month}: {e}")
    time.sleep(2)
```
